```python
import os #to save api key in an environment variable
from langchain_core.prompts import ChatPromptTemplate # to create structured chat prompts.
from langchain_core.output_parsers import StrOutputParser #converts the model's output into a plain Python string.
from langchain_ollama import ChatOllama  #langchain_groq->LangChain integration package specifically for Groq. ChatGroq->is a class that lets LangChain communicate with Groq's chat models.
```

```python
llm=ChatOllama(
model="gemma2:2b", 
temperature=0.1,
)
```

```python
parser=StrOutputParser() #converts the model's output into a plain Python string.
```

```python
def chat():
    chat_history=[
        ("system", "You are a helpful assistant that translates English to French."),
    ]
    print("Chatbot , type 'exit' to end the conversation.")
    while True:
        user_input=input("User: ").strip() #takes the input and removes whitespace from front and end
        if user_input.lower() == "exit":
            print("Exiting the chat. Goodbye!") 
            break
        chat_history.append(("user", user_input))

        #build prompt dynamically
        prompt=ChatPromptTemplate.from_messages(chat_history)   
        chain=prompt | llm | parser  #creates a chain of operations: prompt -> llm -> parser
        response=chain.invoke({}) #invokes the chain and gets the response from the model
        print(f"Chatbot: {response}")
        chat_history.append(("assistant", response)) #adds the model's response to the chat history
        print("-"*80)
```

```python
chat()
```
