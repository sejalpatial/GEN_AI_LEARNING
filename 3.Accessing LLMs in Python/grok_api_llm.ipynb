```python
from groq import Groq
client = Groq()
```

```python
chat_completion=client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Write a short story about a ai failing in future"}
        ]
)
print(chat_completion.choices[0].message.content)
```

```python
chat_completion=client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Write a short story about a ai failing in future"},
        {"role": "assistant", "content": "The Rise and Fall of Echo"},
        {"role":"user","content":"how would you rate this story if you were echo"}
        ]
)
print(chat_completion.choices[0].message.content)
```
