```python
import ollama
response=ollama.chat(
    model="gemma2:2b",
    messages=[
              {"role":"system","content":"You are a helpful assistant"},
              {"role":"user","content":"heyy"}
             ]
)
print(response["message"]["content"])
    
    
```

```python
import ollama

stream=ollama.chat(
    model="gemma2:2b",
    messages=[
        {"role":"system","content":"You are a helpful assistant"},
        {"role":"user","content":"anything of your choice"}
    ],
    stream=True
)
for chunk in stream:
    print(chunk["message"]["content"],end="",flush=True) #using stream the answer appear in chunks like how gpt responds ex; hello , hello there , hello there how .. so one by one word is printed
    #Streaming means the model sends the generated response incrementally (in small chunks or tokens) as it is produced, instead of waiting for the complete response before sending it.
```
