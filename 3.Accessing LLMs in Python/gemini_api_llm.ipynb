```python
from google import genai
```

```python
client = genai.Client()
```

```python
#one time response and doesn't remember the context of the conversation
response=client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Write a short story about a robot learning to love."
)
print(response.text)
```

```python
#it has memory so it can remember the context of the conversation and respond accordingly. It can also generate text in multiple languages and can be used for a variety of applications such as chatbots, content generation, and more.
chat=client.chats.create(model="gemini-2.5-flash")
response=chat.send_message("heyy i m sad")
print(response.text)

response=chat.send_message("how to deal with it")
print(response.text)
```
