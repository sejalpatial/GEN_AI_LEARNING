```python
import os #save the api key

from langchain_chroma import Chroma  #load the vector db
from langchain_huggingface import HuggingFaceEmbeddings #to convert prompt to vector form 
from langchain_groq import ChatGroq
from langchain_classic.chains import RetrievalQA# to wire together the retriever (from Chroma) and LLM (ChatGroq)
from langchain_core.prompts import PromptTemplate #to structure how documents and questions are combined 
```

```python
os.environ["GROQ_API_KEY"]=""
```

```python
vector_db_path = "./vector_db"  #where to store the vector database.
collection_name  = "document_collection"
```

```python
embedding=HuggingFaceEmbeddings()
```

```python
llm =ChatGroq(
    model='llama-3.1-8b-instant',
    temperature=0.0
)
```

```python
vector_store=Chroma(
    collection_name=collection_name,
    embedding_function=embedding,
    persist_directory=vector_db_path
)
```

```python
retriever=vector_store.as_retriever()
```

```python
qa_chain=RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriever,
    return_source_documents=True
)
```

```python
docs = retriever.invoke("What is the document about?")

print("Number of documents:", len(docs))

for doc in docs:
    print(doc.page_content[:500])
    print("---")
```

```python
query="What the document is about"
response=qa_chain.invoke({"query":query})
print(response["result"])
```
