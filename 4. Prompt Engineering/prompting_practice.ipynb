```python
import os
from llama_index.llms.ollama import Ollama
from typing import List, Dict
```

```python
def zero_shot(context:str , prompt:str)->str:
    llm=Ollama(model="gemma2:2b",temperature=1)
    ans=llm.complete(prompt=prompt)
    return ans
```

```python
def few_shot(context:str , prompt:str, examples:List[Dict[str,str]])->str:
    llm=Ollama(model="gemma2:2b",temperature=1)
    ans=llm.complete(prompt=prompt)
    return ans
```

```python
context="Machine Learning is a subset of Artificial Intelligence that enables computers to learn patterns from data without being explicitly programmed. It includes supervised learning, unsupervised learning, and reinforcement learning. Machine Learning is widely used in spam detection, recommendation systems, fraud detection, and medical diagnosis."
prompt="What is Machine Learning?"
zero_shot_response=zero_shot(context=context, prompt=prompt)
print("Zero-Shot Response:", zero_shot_response)
```
